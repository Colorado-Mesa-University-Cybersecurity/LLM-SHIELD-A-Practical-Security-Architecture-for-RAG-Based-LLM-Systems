```python
import os
from dotenv import load_dotenv
import requests
```

```python
load_dotenv()
SUPABASE_URL = os.getenv("SUPABASE_URL")
SUPABASE_KEY = os.getenv("SUPABASE_KEY")
EMAIL = "some@outlook.com"
PASSWORD = "GSV7Or8NurUy9vI1"
USER_ID = "7300f2d0-c562-447e-aa15-316ec3db7258"

# sign in
auth_url = f"{SUPABASE_URL}/auth/v1/token?grant_type=password"
r = requests.post(auth_url, headers={"apikey": SUPABASE_KEY, "Content-Type": "application/json"},
                  json={"email": EMAIL, "password": PASSWORD})
r.raise_for_status()
access_token = r.json()["access_token"]
```

```python
# query table
rest_url = f"{SUPABASE_URL}/rest/v1/renter_transactions?user_id=eq.{USER_ID}&select=*"
headers = {"apikey": SUPABASE_KEY, "Authorization": f"Bearer {access_token}"}
resp = requests.get(rest_url, headers=headers)
rows = resp.json()
if not rows:
    print("No transactions for this user")
else:
    print("Transactions:", rows)
```
